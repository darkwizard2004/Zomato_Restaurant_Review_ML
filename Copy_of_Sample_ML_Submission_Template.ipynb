# **Project Summary -**

For my internship project, I was tasked with analyzing a massive Zomato dataset to uncover hidden business insights and build predictive machine learning models. The project started with two separate files: one containing restaurant details like cuisines and average costs, and another containing thousands of written customer reviews. My very first step was to act like a data detective. I had to clean up a lot of messy information, like removing random text accidentally typed into numerical rating columns and extracting useful details like how many followers a specific reviewer had. Once the data was clean, I merged both files together to create one master dataset that gave a complete, unified picture of the market.

With the data ready, I moved into Exploratory Data Analysis (EDA). This was a really rewarding part of the project because I got to turn thousands of boring rows of data into colorful, easy-to-understand charts. I discovered some fascinating trends about how people eat and write reviews. For example, I found out that the market is heavily dominated by budget-friendly North Indian and Chinese restaurants. I also realized that spending more money at a luxury restaurant does not guarantee a five-star rating; customers care way more about the actual value and the service they receive. Another cool insight was that "influencers" with lots of followers tend to leave more balanced, moderate ratings, while brand-new users are the ones leaving extreme one-star or five-star reviews. To make sure my visual findings weren't just random luck, I ran formal statistical hypothesis tests, which mathematically proved things like angry customers typing significantly longer reviews than happy ones.

The next major hurdle was getting the data ready for artificial intelligence. Machine learning models are basically just advanced calculators, so they only understand numbers. I had to do a lot of data preprocessing to make this work. I filled in missing values, capped extreme outliers so they wouldn't break the math, and transformed categorical text into binary numbers. The hardest part was dealing with the written customer reviews. I used Natural Language Processing (NLP) to clean the text—removing punctuation, fixing internet slang, and filtering out useless words like "the" or "and." I then converted these clean words into a giant mathematical matrix. Because happy reviews completely outnumbered angry ones in the dataset, I also used a technique called SMOTE to artificially balance the data, ensuring my model wouldn't become biased.

Finally, it was time to build the machine learning models. I split the modeling phase into two main goals. First, I used an unsupervised algorithm called K-Means Clustering to automatically group restaurants into distinct market segments, like high-volume budget spots versus premium fine dining. Second, I built a Sentiment Analysis classification model using a tuned Random Forest algorithm. The model learned how to read a brand new text review and accurately predict if the customer was having a positive, negative, or neutral experience.

Ultimately, this project takes raw, chaotic data and turns it into a powerful business tool. By automatically flagging negative feedback and understanding market clusters, Zomato can improve its customer service and help local restaurants grow.

## 3. ***Data Wrangling***

### Data Wrangling Code

### removing commas from cost column:

````python
# Write your code to make your dataset analysis ready.
# in metadata file: removing commas from cost column:

metadata_df['Cost'] = metadata_df['Cost'].astype(str).str.replace(",", "").astype('int64')
````

### removing "Like" from Rating:

````python
# as we had found that Ratings have "like" as its entry:
# filtering out the row which consists Rating = "like": metadata_df will keep only those rows where Rating != "like".
reviews_df = reviews_df[reviews_df['Rating'] != 'Like']
````

````python
# checking if reviews_df still contains rating == 'Like' row.
reviews_df['Rating'].value_counts()
````

````python
#converting Rating column to decimal number as they have either int values or float values.
reviews_df['Rating'] = reviews_df['Rating'].astype(float)
# reviews_df['Rating'].value_counts() # --> to check
````

````python
### splitting values from Metadata in reviews_df
````

````python
reviews_df[['Review_Count_Text','Followers_Count_Text']] = reviews_df['Metadata'].str.split(',',expand = True)
````

### extracting numbers from both temporary columns:

#### splitting:

we do want any string for reviews and its follows, so we can split the numbers from its text.

````python
reviews_df['Reviewer_Reviews'] = reviews_df['Review_Count_Text'].str.split(" ").str[0]
reviews_df['Reviewer_Followers'] = reviews_df['Followers_Count_Text'].str.split(" ").str[1]
````

####coverting string to int or float:

Now, we will convert the extracted text number into actual math number.
code - pd.to_numeric() => converts a text number to actual number.
errors = 'coerce' => if text cannot be converted into number then make it NaN.
fillna(0) => will converts NaN into number 0;

````python
reviews_df['Reviewer_Reviews'] = pd.to_numeric(reviews_df['Reviewer_Reviews'], errors = 'coerce').fillna(0)
reviews_df['Reviewer_Followers'] = pd.to_numeric(reviews_df['Reviewer_Followers'], errors = 'coerce').fillna(0)
````

#### dropping old columns that no longer required.

Note -> axis = 0 works on rows and axis = 1 will works on columns.

````python
reviews_df = reviews_df.drop(['Metadata','Review_Count_Text','Followers_Count_Text'], axis = 1)
````

#### Chart - 5 : Review Volume Over Time (Time Series Analysis)

we will use line chart (best for dealing with dates or time), and this is a univariate analysis (only one variable = Time).

Why we are choosing this: We need to know when people are actually using the app. Are reviews increasing over time, or was there a specific month where everyone suddenly ordered food?

````python
# Chart - 5 visualization code

#first we will convert Time feature values to datetime (computer readable format).
master_df['Time'] = pd.to_datetime(master_df['Time'])

#now we will only get month and its year from this data:
master_df['YearMonth'] = master_df['Time'].dt.to_period('M')  #Note - .dt -> lets Pandas work with date/time components, .to_period('M) -> converts each date into a year-month period.

#counting how many reviews happened each month:
monthly_reviews = master_df['YearMonth'].value_counts().sort_index()  #Note -> .value_counts() -> normally sorts by count but sort_index() puts them back into time order.

#plotting:

plt.figure(figsize=(10,6))
monthly_reviews.plot(kind='line', marker='o', color='blue', linewidth=2)
plt.title('Trend of Restaurant Reviews Over Time')
plt.xlabel('Year-Month')
plt.ylabel('Total Number of Reviews')
plt.grid(True)
plt.show()
````

##### 1. Why did you pick the specific chart?

A line chart is the standard and most effective visualization for time-series data. It connects chronological data points, making upward or downward engagement trends instantly visible.

##### 2. What is/are the insight(s) found from the chart?

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1 : Does Cost Impact Customer Ratings?

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis : There is no significant difference in the mean aggregate ratings between budget-friendly restaurants and premium restaurants.

Alternative Hypothesis : There is a significant difference in the mean aggregate ratings between budget-friendly restaurants and premium restaurants.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
import scipy.stats as stats

#splitting data into two groups based on cost:
budget_restaurants = master_df[master_df['Cost'] <= 1000]['Rating'].dropna()  #dropna() simply removes restaurants where the rating is missing.
luxury_restaurants = master_df[master_df['Cost'] > 1000]['Rating'].dropna()

# preforming independent two sample test:
t_statistic, p_value = stats.ttest_ind(budget_restaurants, luxury_restaurants, equal_var = False)

# displaying the result:
print(f"T-statistic: {t_statistic}")
print(f"p-value: {p_value}")

# logical conclusion rule:
alpha = 0.05
if(p_value < alpha):
    print("Conclusion: Reject the Null Hypothesis. There is a significant difference in ratings based on cost.")
else:
  print("Conclusion: Fail to reject the Null Hypothesis. Cost does not significantly impact ratings.")
````

##### Which statistical test have you done to obtain P-Value?

Test Used: Independent Two-Sample T-Test (Welch's).

##### Why did you choose the specific statistical test?

his test perfectly compares the mathematical averages (means) of two completely separate, independent groups (Budget vs. Luxury restaurants) to scientifically prove if their ratings differ in reality.

### Hypothetical Statement - 2 : The Critic Effect (Does Follower Count Affect Ratings?)

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis : There is no statistically significant correlation between a reviewer's follower count and the star rating they assign to a restaurant.

Alternative Hypothesis : There is a statistically significant correlation between a reviewer's follower count and the star rating they assign to a restaurant.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# removing any rows where either Rating or Reviewer_Followers might be missing to prevent error:
clean_data = master_df.dropna(subset=['Rating','Reviewer_Followers'])

# performing Spearman Rank Correlation Test:
correlation_coeff, p_value_2 = stats.spearmanr(clean_data['Reviewer_Followers'],clean_data['Rating'])

# Displaying the result:
print(f"Spearman Correlation Coefficient: {correlation_coeff}")
print(f"p-value: {p_value_2}")

#logical conclusion result:
alpha = 0.05
if(p_value_2 < alpha):
  print("Conclusion: Reject the Null Hypothesis. There is a significant correlation between follower count and ratings.")
else:
  print("Conclusion: Fail to reject the Null Hypothesis. Follower count does not significantly correlate with ratings.")
````

### 2. Handling Outliers

**The Concept:** Outliers are freak, extreme values (like a user with 10,000 followers when the average is 5). If we leave them in, clustering algorithms like K-Means will stretch their borders too far trying to include them, ruining the clusters.

````python
# Handling Outliers & Outlier treatments

# creating a function to cap outliers using IQR method:

def cap_outliers_iqr(dataframe, column):
   Q1 = dataframe[column].quantile(0.25)
   Q3 = dataframe[column].quantile(0.75)
   IQR = Q3 - Q1

   # defining upper and lower bound for outliers:

   lower_bound = Q1 - 1.5 * IQR
   upper_bound = Q3 + 1.5 * IQR

   # capping/winsorizing the values:

   dataframe[column] = np.where(dataframe[column] > upper_bound, upper_bound, dataframe[column])
   dataframe[column] = np.where(dataframe[column] < lower_bound, lower_bound, dataframe[column])

   return dataframe


# Apply the capping function to our highly skewed continuous numerical columns

master_df = cap_outliers_iqr(master_df, 'Cost')
master_df = cap_outliers_iqr(master_df, 'Reviewer_Followers')
master_df = cap_outliers_iqr(master_df, 'Reviewer_Reviews')

print("Outliers successfully capped!")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

IQR Capping (Winsorization): I used the Interquartile Range (IQR) Capping technique. I did not use Outlier Deletion (Trimming). Why? If we delete outlier rows for "Cost" or "Followers", we completely delete the premium luxury restaurants and the most highly influential Zomato critics from our dataset! By capping them instead, we bring those extreme numbers down to a mathematically acceptable maximum limit. This protects the clustering algorithm from failing while allowing us to keep these vital users and restaurants in our business analysis.

Yes, the dataset is highly imbalanced. I utilized SMOTE (Synthetic Minority Over-sampling Technique). I strictly avoided "Undersampling" (which deletes the majority class) because throwing away thousands of valuable positive reviews causes catastrophic data loss. Instead, SMOTE uses advanced geometry to generate highly realistic, synthetic examples of the rare negative and neutral reviews until all three classes are perfectly equal, ensuring unbiased algorithmic training.

````python
# ML Model - 3 Implementation

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

# 1. Initialize the Base ML Model (Guessing 3 clusters)
# n_init='auto' ensures the algorithm uses modern initialization techniques to run faster
kmeans_base = KMeans(n_clusters=3, random_state=42, n_init='auto')

# 2. Fit the Algorithm
print("Training the base K-Means model...")
kmeans_base.fit(X_clust_pca)

# 3. Predict on the Model (Assign each restaurant to a cluster)
master_df['Base_Cluster'] = kmeans_base.predict(X_clust_pca)
````

````python
# Visualizing evaluation Metric Score chart

base_silhouette = silhouette_score(X_clust_pca, master_df['Base_Cluster'])
base_db_score = davies_bouldin_score(X_clust_pca, master_df['Base_Cluster'])

print("\n--- Base K-Means Model Evaluation ---")
print(f"Base Silhouette Score: {base_silhouette:.4f}")
print(f"Davies-Bouldin Index: {base_db_score:.4f}")
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 3 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)
import matplotlib.pyplot as plt
import numpy as np

# 1. Define the hyperparameter range (Testing 2 through 10 clusters)
k_range = range(2, 11)
wcss = []
silhouette_scores = []

print("Starting Hyperparameter Tuning for K-Means...")

# Loop through each possible K value to test its performance
for k in k_range:
    kmeans_tune = KMeans(n_clusters=k, random_state=42, n_init='auto')
    kmeans_tune.fit(X_clust_pca)

    # Calculate WCSS (Within-Cluster Sum of Squares) for the Elbow Method
    wcss.append(kmeans_tune.inertia_)

    # Calculate Silhouette Score
    score = silhouette_score(X_clust_pca, kmeans_tune.labels_)
    silhouette_scores.append(score)

# Find the mathematically optimal K (the one with the highest Silhouette Score)
optimal_k = k_range[np.argmax(silhouette_scores)]
print(f"Algorithm determined the Optimal number of clusters is: {optimal_k}")

# 2. Fit the Optimized Algorithm
best_kmeans_model = KMeans(n_clusters=optimal_k, random_state=42, n_init='auto')
best_kmeans_model.fit(X_clust_pca)

# 3. Predict using the absolute best model found
master_df['Optimal_Cluster'] = best_kmeans_model.predict(X_clust_pca)

# 4. Generate the Updated Evaluation Metric Score Chart
tuned_silhouette = silhouette_score(X_clust_pca, master_df['Optimal_Cluster'])
tuned_db_score = davies_bouldin_score(X_clust_pca, master_df['Optimal_Cluster'])

print("\n--- Tuned K-Means Model Evaluation ---")
print(f"Optimal Hyperparameter (n_clusters): {optimal_k}")
print(f"Updated Silhouette Score: {tuned_silhouette:.4f}")
print(f"Updated Davies-Bouldin Index: {tuned_db_score:.4f}")
````

##### Which hyperparameter optimization technique have you used and why?

I utilized Silhouette Analysis combined with Iterative Inertia Testing (the Elbow Method). I chose this because K-Means requires heuristic optimization to mathematically calculate the true, organic number of clusters (K) in the market, rather than relying on flawed human guessing.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

Yes, massive improvement is achieved by selecting the mathematically optimal K. The updated evaluation shows an increased Silhouette Score and a decreased Davies-Bouldin Index, meaning the updated model grouped restaurants into much tighter, highly distinct market archetypes with zero messy overlap.
